```python
class Employee:
    def __init__(self):
        self.name = "Rahul"        #Public
        self._salary = 50000       #Protected
        self._password = "12345"    #Private
    def show_details(self):
        print(self.name)
        print(self._salary)
        print(self._password)
emp = Employee()

print(emp.name)       #Works
print(emp._salary)    #Works, but discouraged
print(emp._password)  #Error

emp.show_details()
```

```python
class Employee:
    def __init__(self):
        self.name = "Anu"        #Public
        self.__salary = 50000       #Protected
        self.__password = "12345"    #Private

    def show_parent(self):
        print("Name: ",self.name)
        print("Salary: ",self.__salary)
        print(self.__password)

class Manager(Employee):
  def show_manager(self):
    print("Name: ",self.name)             #Public -> accessible
    print("Salary: ",self._salary)       #Protected -> accessible
    #print("Password: ",self.__password)   #Private -> NOT accessible

manager = Manager()
manager.show_manager()
```
